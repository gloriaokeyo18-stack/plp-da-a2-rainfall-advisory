
# Rainfall Analysis and Planting Advisory

## Project Overview

This project analyses monthly rainfall data across five counties using Python and NumPy. It identifies rainfall patterns, annual totals, wettest months, dry months, and recommended planting windows to support agricultural planning by an NGO.

## Objectives

- Clean and validate rainfall data.
- Perform numerical analysis using NumPy.
- Identify suitable two-month planting windows.
- Generate a CSV summary for reporting and decision-making.


```python
import numpy as np
import csv
```

```python
months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

machakos_raw = [
    "48", "35", "88", "N/A", "140", " 20 ",
    "9", "12", "", "45", "182", "96"
]
```

```python

def clean_machakos(data, months):

    valid_values = []
    invalid_indices = []

    # Step 1: Convert valid rainfall readings to integers
    for i, value in enumerate(data):
        try:
            valid_values.append(int(value.strip()))

        except ValueError:
            invalid_indices.append(i)

    # Step 2: Calculate the rounded average of valid values
    average_rainfall = round(np.mean(valid_values))

    # Step 3: Replace invalid values with the average
    cleaned_values = []

    for i, value in enumerate(data):
        try:
            cleaned_values.append(int(value.strip()))

        except ValueError:
            cleaned_values.append(average_rainfall)

    # Step 4: Print the months that were imputed
    print("Imputed months:")

    for i in invalid_indices:
        print(f"{months[i]}: {average_rainfall} mm")

    return cleaned_values


machakos_clean = clean_machakos(machakos_raw, months)

print("\nCleaned Machakos rainfall data:")
print(machakos_clean)
```

```python
import numpy as np

months = [
    "January", "February", "March", "April",
    "May", "June", "July", "August",
    "September", "October", "November", "December"
]

machakos_raw = [
    "48", "35", "88", "N/A", "140", " 20 ",
    "9", "12", "", "45", "182", "96"
]

print("Raw data loaded successfully!")
print("Number of months:", len(months))
print("Number of rainfall readings:", len(machakos_raw))
```

```python
counties = ["Nairobi", "Nakuru", "Kisumu", "Mombasa", "Machakos"]

clean_rows = [

    [64, 56, 92, 219, 176, 35, 18, 24, 31, 61, 165, 104],  # Nairobi

    [35, 42, 70, 145, 120, 78, 96, 105, 72, 55, 60, 38],   # Nakuru

    [58, 90, 150, 240, 180, 95, 70, 85, 90, 88, 130, 95],  # Kisumu

    [30, 15, 55, 150, 240, 110, 90, 65, 60, 90, 100, 70]   # Mombasa

]

print("Counties loaded:", counties)
print("Number of counties with clean data:", len(clean_rows))
```

```python
# Combine rainfall data from all five counties
all_rainfall = clean_rows + [machakos_clean]

# Convert the combined data into a NumPy array
rainfall_array = np.array(all_rainfall)

# Print the array
print("Rainfall data:")
print(rainfall_array)

# Confirm its shape and data type
print("\nArray shape:", rainfall_array.shape)
print("Array data type:", rainfall_array.dtype)
```

Cell 6: Annual Rainfall totals

```python
# Calculate annual rainfall totals for each county
annual_totals = np.sum(rainfall_array, axis=1)

# Display results
print("Annual Rainfall Totals:")

for county, total in zip(counties, annual_totals):
    print(f"{county}: {total} mm")
```

Cell 7: Monthly rainfall averages

```python
# Calculate average rainfall for each month across all counties
monthly_averages = np.mean(rainfall_array, axis=0)

# Display results
print("Average Monthly Rainfall Across All Counties:")

for month, average in zip(months, monthly_averages):
    print(f"{month}: {average:.2f} mm")
```

Cell 8: Identify wettest month


```python

# Find the index of the wettest month for each county
wettest_month_indices = np.argmax(rainfall_array, axis=1)

# Match county names with their wettest month indices
print("Wettest Month for Each County:")

for county, index in zip(counties, wettest_month_indices):
    print(f"{county}: {months[index]}")
```

Cell 9: Identify and count dry months


```python

# Create a Boolean mask for rainfall below 50 mm
dry_mask = rainfall_array < 50

# Count the number of dry months for each county
dry_month_counts = np.sum(dry_mask, axis=1)

# Display results
print("Number of Dry Months per County:")

for county, count in zip(counties, dry_month_counts):
    print(f"{county}: {count} dry months")
```

Cell 10: Create the shifted array

```python
# Calculate combined rainfall for every pair of consecutive months
planting_rainfall = (
    rainfall_array[:, :-1] + rainfall_array[:, 1:]
)

# Display the combined rainfall array
print("Combined rainfall for consecutive months:")
print(planting_rainfall)

# Confirm the shape
print("\nPlanting rainfall array shape:", planting_rainfall.shape)
```

Cell 11: Find the best planting window

```python
# Find the index of the two-month window with the highest rainfall
planting_window_indices = np.argmax(planting_rainfall, axis=1)

# Find the highest combined rainfall for each county
highest_window_rainfall = np.max(planting_rainfall, axis=1)

# Display the recommended planting windows
print("Recommended Planting Windows:")

for county, index, total in zip(
    counties,
    planting_window_indices,
    highest_window_rainfall
):
    window = f"{months[index]} - {months[index + 1]}"
    print(f"{county}: {window} ({total} mm)")
```

Cell 12: Export the rainfall summary

```python
import csv

print("CSV module imported successfully!")
print("CSV module location:", csv.__file__)
print("CSV writer available:", hasattr(csv, "writer"))
```

```python

# Create and write the rainfall summary CSV file

with open("rainfall_summary.csv", "w", newline="") as file:

    writer = csv.writer(file)

    # Write the column headers
    writer.writerow([
        "county",
        "annual_total_mm",
        "wettest_month",
        "dry_months",
        "planting_window"
    ])

    # Write one row for each county
    for county, annual, wet_idx, dry_count, plant_idx in zip(
        counties,
        annual_totals,
        wettest_month_indices,
        dry_month_counts,
        planting_window_indices
    ):

        wettest_month = months[wet_idx]

        planting_window = (
            f"{months[plant_idx]}-{months[plant_idx + 1]}"
        )

        writer.writerow([
            county,
            annual,
            wettest_month,
            dry_count,
            planting_window
        ])

print("rainfall_summary.csv created successfully!")
```

```python
with open("rainfall_summary.csv", "r") as file:
    reader = csv.reader(file)
    for row in reader:
        print(row)
```

### Planting Advisory

The rainfall analysis indicates that April–May is the strongest historical two-month rainfall window for Nairobi, Nakuru, Kisumu and Mombasa, while Machakos records its highest combined rainfall in November–December. The NGO can use these county-specific periods to guide initial planting advisories and farmer communication. However, farmers should confirm local rainfall onset and consider short-term weather forecasts before planting, rather than relying solely on historical rainfall totals. In areas experiencing several dry months, particularly Machakos, the NGO should also promote soil-moisture conservation and appropriate water-management practices. These findings represent historical observations from the supplied dataset and should not be interpreted as rainfall forecasts or guarantees of successful crop establishment.

### Data Quality Notes

The Machakos rainfall dataset contained two invalid entries: `N/A` for April and a blank value for September. These entries were replaced with the mean of the ten valid monthly observations, which was 67.5 mm and was rounded to 68 mm. This imputation allowed the five counties to be analysed using a complete 5 × 12 rainfall array. However, replacing missing observations with the mean may affect the calculated annual total, monthly averages and planting window for Machakos. The results should therefore be interpreted with caution, particularly because the analysis uses only one year's monthly rainfall observations.
